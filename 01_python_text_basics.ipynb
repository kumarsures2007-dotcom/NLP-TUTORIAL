# Topic : Introduction  of Text Data



NLP works with human language.
In Python, text is mainly represented using strings.
A string is a sequence of characters.

```python
sentence = "I am Learning NLP"
word = "Python"
name = "Sureshkumar"
paragraph = """ computer technology that allows 
            machines to perform tasks that normally need human thinking,
            like learning, reasoning, and solving problems. """
```

```python
print("Length of the sentence:", len(sentence))
print("Length of the word:", len(word))
print("Length of the name:", len(name))
print("Length of the paragraph:", len(paragraph))
    
```

```python
import string

space = sentence.count(" ")

print("Spaces in sentence:", space)

punctuation = 0

for char in paragraph:
    if char in string.punctuation:
        punctuation +=1

print("Punctuation marks in paragraph:", punctuation)
```

```python
print(sentence[0:4])
print(sentence[-1])
print(sentence[5:13]) 
```

```python
lower_case = sentence.lower()
upper_case = sentence.upper()

print("Lowercase in sentence:", lower_case)
print("Uppercase in sentence:", upper_case)
```

```python
word = "solving" 

if word in paragraph:
    print("Word is present in paragraph")
else:
    print("Word is not present in the paragraph")
```

```python
position = paragraph.find("reasoning")

print("Position of the word 'reasoning' in paragraph:", position)
```

```python
text = "I Love Python Programming"

new_text = text.replace("Python","Pytorch and Tensorflow")

print("Original Text:", text)

print("New Text:", new_text)
```

```python
sent = "I love Python. Python is easy. Python is powerful."

lower_sent = sent.lower()

count = lower_sent.count("python")

print("Count of the word 'Python' in the sentence:", count)
```

```python
nospace = paragraph.replace(" ", "")
print("Paragraph without spaces:", nospace)
```

```python
split_text = paragraph.split()
print("Split paragraph into words:", split_text)
```

```python
join_text = " ".join(split_text)

print("Join the words back into a paragraph:", join_text)
```

```python
word1 = "python"

word2 = "Python"

if word1.lower() == word2.lower():
    print("The words are equal (case-insensitive comparison).")

else:
    print("The words are not equal.")
```
