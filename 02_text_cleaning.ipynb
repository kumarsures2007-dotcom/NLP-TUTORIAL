## 1. What is Text Cleaning?

Real-world text is often messy.

Example:

"   I LOVE NLP!!! 😊 I am learning NLP in 2026...   "

Text cleaning converts messy text into a more consistent format.

General pipeline:

Raw Text
    ↓
Lowercase
    ↓
Remove unwanted content
    ↓
Normalize whitespace
    ↓
Clean Text

```python
import string 
import re
```

## 2. Create Raw Text

We will intentionally create messy text so that we can practice cleaning it.

```python
raw_text ="  I LOVE NLP!!! 😊 I am learning NLP in 2026...   "
print(raw_text)
```

## 3. Inspect the Raw Text

Before cleaning data, we should understand what is inside it.

We will check:
- Number of characters
- Number of spaces
- Number of words
- Number of digits
- Number of punctuation marks

```python
print("Number of characters in raw_text:", len(raw_text))
print("Number of spaces in raw_text:", raw_text.count(" "))
print("Number of words in raw_text:", len(raw_text.split()))
```

```python
count = 0
for digit in raw_text:
    if digit.isdigit():
        count+=1

print("Number of digits in raw_text:", count) 
```

```python
count = 0
for punctuation in raw_text:
    if punctuation in string.punctuation:
        count+=1

print("Number of punctuation marks in raw_text:", count)
```

## 4. Lowercasing

Lowercasing converts uppercase characters into lowercase characters.

Example:

"I LOVE NLP"

becomes:

"i love nlp"

This is commonly used to make text more consistent.

```python
print("Lower case : ",raw_text.lower())
```

## 5. Removing Leading and Trailing Spaces

Leading spaces appear before the text.

Trailing spaces appear after the text.

Example:

"   Hello World   "

becomes:

"Hello World"

We should not remove spaces between words.

```python
text = "     I am learning NLP     "
cleaned_text  = text.strip()

print("Original text : ",text)

print("Cleaned text : ",cleaned_text)
```

## 6. Removing Punctuation

Punctuation includes characters such as:

! . , ? ; : ' " ( ) [ ]

For some NLP tasks, punctuation can be removed.

However, punctuation should not always be removed.

For example:

"Great!"
"Great?"

may contain different sentiment information.

```python
text = "Hello!!! NLP is amazing... Really?"

cleaned_text = re.sub(r'[^\w\s]', '', text)

print("Original text : ",text)
print("Cleaned text : ",cleaned_text)
```

## 7. Handling Numbers

Numbers may or may not be useful.

Example:

"Python 3.13 was released in 2026."

For some NLP tasks, numbers can be removed.

But for other tasks, numbers are important.

Examples:
- Product names
- Prices
- Dates
- Versions
- Age
- Financial information

```python
text = "Python 3.13 was released in 2026."

text_without_numbers = re.sub(r'\d+','',text)

print("Original text : ",text)  
print("Text without numbers : ",text_without_numbers)
```

## 8. Handling Special Characters

Example:

"Hello @Suresh #NLP $AI %MachineLearning"

Special characters include:

@ # $ % & *

Whether we remove them depends on the NLP task.

```python
text = "Hello @Suresh #NLP $AI %MachineLearning"

clean_text = re.sub(r"[^a-zA-Z0-9\s]", "", text)

print("Original text : ",text)
print("Cleaned text : ",cleaned_text)
```

## 9. Handling URLs

URLs often appear in web and social-media text.

Example:

https://example.com

For sentiment analysis, URLs may not be useful.

For website classification, URLs may be important.

So URL removal depends on the task.

```python
text = "Learn NLP at https://youtube.com"

cleaned_text = re.sub(r"https?://\S+|www\.\S+", "", text)

print("Original text : ",text)
print("Cleaned text : ",cleaned_text)
```

## 10. Handling Email Addresses

Example:

"Contact me at suresh@example.com"

We can detect email addresses using a regular expression.

Whether we remove them depends on the application.

```python
text = "Contact me at kumarsures@gmail.com for NLP resources."

clean_text = re.sub(r"\S+@\S+", "", text)

print("Before:", text)
print("After:", clean_text)
```

## 11. Handling Emojis

Example:

"I love NLP 😊🔥❤️"

Emojis can contain useful sentiment information.

For sentiment analysis, removing emojis may sometimes reduce useful information.

Therefore, we should decide whether to remove them based on the task.

```python
text = "I love NLP 😊🔥❤️"

print(text)
```

## 12. Normalizing Multiple Spaces

Real text can contain:

"I    am     learning       NLP"

We want:

"I am learning NLP"

This is called whitespace normalization.

```python
text = "I    am     learning       NLP"
cleaned_text = re.sub(r'\s+', ' ', text)
print("Before:", text)
print("After:", cleaned_text)
```

## 13. Handling New Lines and Tabs

Text can contain:
- Spaces
- Tabs
- New lines

These are all forms of whitespace.

We can normalize them into regular spaces.

```python
text = "I am learning NLP\nPython\tMachine Learning"

normalized_text = re.sub(r'\s+', ' ', text)

print("Before : ", text)

print("After : ", normalized_text)
```

# 14. Complete Text Cleaning Pipeline

Now we will combine the techniques we learned.

Pipeline:

Raw Text
    ↓
Remove URLs
    ↓
Remove Emails
    ↓
Lowercase
    ↓
Remove punctuation
    ↓
Handle numbers
    ↓
Handle special characters
    ↓
Normalize whitespace
    ↓
Clean Text

```python
def clean_text(text):

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", "", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", "", text)

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

## Testing the Cleaning Function

We should never test an NLP preprocessing function with only one example.

Let's test multiple types of text.

```python
text = "  Please contact our support team immediately at HELP_DESK@company99.com if you experience any system errors, or you can visit our troubleshooting portal directly at status.net for a complete list of live updates. We are working around the clock—24 hours a day, 7 days a week—to resolve these technical issues, so please review the online guide carefully before submitting a new ticket.   "

cleaned = clean_text(text)

print("Original:")
print(text)

print("\nCleaned:")
print(cleaned)
```

## Testing the Cleaning Function

We should never test an NLP preprocessing function with only one example.

Let's test multiple types of text.

```python
texts = [
    "HELLO WORLD!!!",
    "   I am learning Python   ",
    "AI is amazing!!! 2026",
    "Python @ NLP #MachineLearning",
    "Visit https://example.com",
    "Contact me at student@gmail.com"
]

for text in texts:
    print("Original :", text)
    print("Cleaned  :", clean_text(text))
    print("-" * 60)
```

## Before vs After

We will compare:
- Character count
- Word count
- Text appearance

```python
text = "   WOW!!! NLP is AMAZING 😊😊 in 2026!!!   "

cleaned = clean_text(text)

print("Original text:", text)
print("Cleaned text:", cleaned)

print("\nOriginal characters:", len(text))
print("Cleaned characters:", len(cleaned))

print("\nOriginal words:", len(text.split()))
print("Cleaned words:", len(cleaned.split()))
```

# 18. Real NLP Example

Let's use a realistic piece of text.

We will inspect the text and clean it.

```python
paragraph = """
   Artificial Intelligence is AMAZING!!! 🤖
   NLP helps computers understand human language.
   Learn more at https://example.com
   Contact us at ai@example.com
   Course year: 2026
"""
```

```python
def clean_text(paragraph):

    # Remove URLs
    paragraph = re.sub(r"https?://\S+|www\.\S+", "", paragraph)

    # Remove email addresses
    paragraph = re.sub(r"\S+@\S+", "", paragraph)

    # Convert to lowercase
    paragraph = paragraph.lower()

    # Remove punctuation
    paragraph = paragraph.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    paragraph = re.sub(r"\d+", "", paragraph)

    # Normalize whitespace
    paragraph = re.sub(r"\s+", " ", paragraph).strip()

    return paragraph


print("Original paragraph:")
print(paragraph)
print("=" * 60)

print("Cleaned paragraph:")
print(clean_text(paragraph))
```

# 19. Important NLP Decision

Text preprocessing is NOT the same for every NLP problem.

For example:

Sentiment Analysis
→ Emojis may be useful.

Financial NLP
→ Numbers and currency may be important.

Social Media Analysis
→ Hashtags and mentions may be important.

Website Classification
→ URLs may be important.

Therefore:

Do not blindly remove everything.

Always ask:

"What information does this text contain, and what does my NLP task need?"
