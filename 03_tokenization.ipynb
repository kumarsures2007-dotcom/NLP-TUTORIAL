## 1. What is Tokenization?

Tokenization means breaking text into smaller meaningful units called tokens.

Example:

"I love Python"

can become:

"I"
"love"
"Python"

These individual pieces are called tokens.

Tokenization is an important preprocessing step in NLP.

"I love NLP"
      ↓
["I", "love", "NLP"]

```python
sentence = "I Love Python "

print(sentence)
```

## 2. Character Tokenization

One approach is to treat every character as a token.

Example:

"Python"

can become:

P
y
t
h
o
n

This is called character-level tokenization.

```python
words = "Python"

for char in words:
    print(char)
```

```python
for i in range(0,len(words)):
    print(words[i],i)
```

## 4. Word Tokenization

Word tokenization breaks text into words.

Example:

"I love Natural Language Processing"

becomes:

["I", "love", "Natural", "Language", "Processing"]

Each item is a word token.

```python
sentence = "I love machine learning and natural language processing"

words = sentence.split()

print(words)

print("="*50)

print("Length of words list:", len(words))
```

# 4. Whitespace Tokenization

The simplest word tokenization method is splitting text based on whitespace.

Whitespace includes:

- Normal spaces
- Tabs
- New lines

Example:

"I love NLP"

can become:

["I", "love", "NLP"]

```python
text = "I    love    NLP"

tokens = text.split()

print(tokens)
```

```python
text = "I\tlove\tNLP"

tokens = text.split()

print(tokens)
```

```python
text = """I
love
NLP"""

tokens = text.split()

print(tokens)
```

# 5. Tokenization and Punctuation

Consider:

"I love Python!"

Simple whitespace splitting produces:

["I", "love", "Python!"]

Notice that "Python!" contains punctuation.

A more advanced tokenizer may separate:

["I", "love", "Python", "!"]

Punctuation handling depends on the tokenizer and NLP task.

```python
sentences = [
    "Hello, world!",
    "Python is amazing!!!",
    "Are you learning NLP?",
    "I love AI, ML and Python."
]

for sentence in sentences:
    tokens = sentence.split()
    print(sentence)
    print(tokens)
    print()
```

# 6. Sentence Tokenization

Sentence tokenization separates a paragraph into individual sentences.

Example:

"I love Python. Python is easy. NLP is interesting."

becomes:

Sentence 1 → "I love Python."
Sentence 2 → "Python is easy."
Sentence 3 → "NLP is interesting."

```python
text  = "I love Python. Python is easy. NLP is interesting."

for sentence in text.split('.'):
    tokens = sentence.split()
    print(sentence)
    print(tokens)
  
```

```python
print("Number of sentences:", len(sentences))
```

# 7. Problem with Simple Sentence Splitting

Simple splitting on "." does not always work.

Example:

"Dr. APJ Abdul Kalam was a scientist. He inspired many students."

If we split using ".", we may incorrectly consider:

"Dr"

as a complete sentence.

Real NLP tokenizers use more advanced linguistic rules.

```python
paragraph = "Dr. APJ Abdul Kalam was a scientist. He inspired many students."

sentences = paragraph.split(".")

for sentence in sentences:
    print(sentence.strip())
```

# 9. Unique Tokens

Example:

"I love Python and I love NLP"

Tokens:

Python

Unique tokens:

Python
and

Vocabulary size = 5

```python
text = "I love Python and I love NLP"

tokens = text.split()

unique_tokens = set(tokens)

print("Tokens : ", tokens)
print("Unique Tokens : ", unique_tokens)
```

# 11. Token Frequency

Token frequency tells us how many times each token appears.

Example:

"I love Python. Python is easy. Python is powerful."

Python appears multiple times.

We can use a dictionary to count token frequencies.

```python
text = "I love Python Python is easy Python is powerful"

tokens = text.split()

frequency = {}

for token in tokens:
    if token in frequency:
        frequency[token] += 1
    else:
        frequency[token] = 1

print("Token Frequencies : ", frequency)
```

```python
print("Python frequency:", frequency.get("Python", 0))
print("I frequency:", frequency.get("I", 0))
print("love frequency:", frequency.get("love", 0))
```

# 12. Case Sensitivity

Consider:

"Python python PYTHON Python"

A computer can treat these as different tokens.

Python
python
PYTHON

If we convert everything to lowercase:

python
python
python
python

They become the same token.

Therefore, cleaning and tokenization are connected.

```python
text = "Python python PYTHON Python"

tokens = text.split()

print("Original tokens:", tokens)
print("Unique tokens:", set(tokens))
```

```python
lower_text = text.lower()

tokens = lower_text.split()

print("Lowercase tokens:", tokens)
print("Unique tokens:", set(tokens))
```

# 13. Cleaning + Tokenization

We can combine Day 2 and Day 3.

Raw text
    ↓
Text cleaning
    ↓
Lowercase
    ↓
Whitespace normalization
    ↓
Tokenization
    ↓
Tokens

This is a basic NLP preprocessing pipeline.

```python
import re
import string 
def clean_text(text):

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", "", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", "", text)

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
text = "   I LOVE NLP!!! NLP is AMAZING!!! 😊   "

cleaned = clean_text(text)

tokens = cleaned.split()

print("Original:")
print(text)

print("\nCleaned:")
print(cleaned)

print("\nTokens:")
print(tokens)
```

# 14. NLP Library Tokenization

Python's split() is useful for learning.

However, real NLP applications often use specialized tokenizers.

These tokenizers can better handle:

- Punctuation
- Contractions
- Special characters
- Sentence boundaries
- Language-specific rules

We will now compare simple splitting with an NLP tokenizer.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

sentence = "I love Python!"

tokens = word_tokenize(sentence)

print(tokens)
```

```python
sentence = "Hello, world!"

print("Using split():")
print(sentence.split())

print("\nUsing NLP tokenizer:")
print(word_tokenize(sentence))
```

```python
from nltk.tokenize import sent_tokenize

paragraph = """Artificial Intelligence is changing many industries.
Natural Language Processing helps computers understand human language.
Students can use NLP to build intelligent applications."""

sentence = sent_tokenize(paragraph)

print("Number of sentences:", len(sentence))

print("Sentence : ",sentence)
```

# 16. Real-World NLP Example

We will analyze this paragraph:

"Artificial Intelligence is changing many industries.
Natural Language Processing allows computers to work with human language.
Students can use NLP to build intelligent applications."

Tasks:

1. Sentence tokenize
2. Word tokenize
3. Count sentences
4. Count total tokens
5. Find unique tokens
6. Find vocabulary size

```python
paragraph = """
Artificial Intelligence is changing many industries.
Natural Language Processing allows computers to work with human language.
Students can use NLP to build intelligent applications.
"""

sentences = sent_tokenize(paragraph)

print("Number of sentences:", len(sentences))

for i, sentence in enumerate(sentences, start=1):
    print(f"\nSentence {i}:")
    print(sentence)
```

```python
all_tokens = []

for sentence in sentences:
    tokens = word_tokenize(sentence)
    all_tokens.extend(tokens)

print("Total tokens:", len(all_tokens))
print("Tokens:")
print(all_tokens)
```

```python
unique_tokens = set(all_tokens)

print("Unique tokens:", unique_tokens)
print("Vocabulary size:", len(unique_tokens))
```
